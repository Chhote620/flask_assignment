```python
#Q1. Write a code to reverse a string.
# Ans:-
def reverse_string(string):
    reversed_string = ""
    for i in range(len(string) - 1, -1, -1):
        reversed_string += string[i]
    return reversed_string
```

```python
#Q2. Write a code to count the number of vowels in a string.
# Ans:-
def count_vowels(string):
    vowels = "aeiouAEIOU"
    count = 0
    for char in string:
        if char in vowels:
            count += 1
    return count
```

```python
#Q3. Write a code to check if a given string is a palindrome or not.
# Ans:-
def is_palindrome(string):
    reversed_string = ""
    for i in range(len(string) - 1, -1, -1):
        reversed_string += string[i]
```

```python
#Q7. Write a code to determine if a string has all unique characters.
# Ans:-
def has_unique_characters(string):
    unique_chars = set()
    for char in string:
        if char in unique_chars:
            return False
        unique_chars.add(char)
    return True
```

```python
#Q8. Write a code to convert a given string to uppercase or lowercase.
# Ans:-
def convert_case(string, case):
    if case == "uppercase":
        return string.upper()
    elif case == "lowercase":
        return string.lower()
    else:
        return string
```

```python
#Q9. Write a code to count the number of words in a string.
# Ans:-
def count_words(string):
    words = string.split()
    return len(words)
```

```python
#Q10. Write a code to concatenate two strings without using the + operator.
# Ans:-
def concatenate_strings(string1, string2):
    result = ""
    for char in string1:
        result += char
    for char in string2:
        result += char
    return result
```

```python
#Q14. Write a code to reverse a list in-place without using any built-in reverse functions.
# Ans:-
def reverse_list(lst):
    start = 0
    end = len(lst) - 1
    while start < end:
        lst[start], lst[end] = lst[end], lst[start]
        start += 1
        end -= 1
```

```python
#Q15. Implement a code to find and remove duplicates from a list while preserving the original order of elements.
# Ans:-
def remove_duplicates(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            result.append(item)
            seen.add(item)
    return result

original_list = [1, 2, 2, 3, 4, 1, 5, 3]
clean_list = remove_duplicates(original_list)
print("List after removing duplicates:", clean_list)



```

```python
#Q20. Write a code to shuffle a given list randomly without using any built-in shuffle functions.
# Ans:-
import random

def manual_shuffle(lst):
    n = len(lst)
    for i in range(n - 1, 0, -1):
        j = random.randint(0, i)

        lst[i], lst[j] = lst[j], lst[i]
    return lst

original_list = [1, 2, 3, 4, 5]
shuffled_list = manual_shuffle(original_list.copy())
print("Shuffled List:", shuffled_list)





```

```python
#Q21. Write a code that takes two tuples as input and returns a new tuple containing elements that are common to both input tuples.
# Ans:-
def find_common_elements(tuple1, tuple2):
    common_elements = []
    for element in tuple1:
        if element in tuple2:
            common_elements.append(element)
    return tuple(common_elements)
```

```python
#Q27. Develop a code that takes a tuple of integers as input. The function should return the maximum and minimum values from the tuple using tuple unpacking.
# Ans:-
def find_max_min(numbers):
    max_val = max(numbers)
    min_val = min(numbers)
    return (max_val, min_val)  # Returning a tuple

# Example usage
input_str = input("Enter integers separated by commas: ")
input_tuple = tuple(map(int, input_str.split(',')))

# Tuple unpacking
maximum, minimum = find_max_min(input_tuple)

print("Maximum value:", maximum)
print("Minimum value:", minimum)

```

```python
#Q31. Write a code that takes a list of words as input and returns a dictionary where the keys are unique words and the values are the frequencies of those words in the input list.
# Ans:-
def word_frequencies(word_list):
    freq_dict = {}
    for word in word_list:
        if word in freq_dict:
            freq_dict[word] += 1
        else:
            freq_dict[word] = 1
    return freq_dict

# Example usage:
words = input("Enter a list of words (space-separated): ").split()
result = word_frequencies(words)
print("Word Frequencies:", result)

```

```python
#Q32. Write a code that takes two dictionaries as input and merges them into a single dictionary. If there are common keys, the values should be added together
# Ans:-
def merge_dictionaries(dict1, dict2):
    merged = dict1.copy()  # Start with a copy of the first dictionary
    for key, value in dict2.items():
        if key in merged:
            merged[key] += value  # Add values for common keys
        else:
            merged[key] = value  # Add new key-value pairs
    return merged

# Example usage:
dict1 = {'a': 10, 'b': 20, 'c': 30}
dict2 = {'b': 5, 'c': 15, 'd': 25}

result = merge_dictionaries(dict1, dict2)
print("Merged Dictionary:", result)

```

```python
#Q34. Write a code that takes a dictionary as input and returns a sorted version of it based on the values. You can choose whether to sort in ascending or descending order.
# Ans:-
def sort_dict_by_values(input_dict, descending=False):
    # Sort dictionary by values
    sorted_dict = dict(sorted(input_dict.items(), key=lambda item: item[1], reverse=descending))
    return sorted_dict

# Example usage:
data = {'apple': 3, 'banana': 1, 'cherry': 5, 'date': 2}

# Sort in ascending order
ascending_sorted = sort_dict_by_values(data)
print("Ascending Order:", ascending_sorted)

# Sort in descending order
descending_sorted = sort_dict_by_values(data, descending=True)
print("Descending Order:", descending_sorted)

```

```python
#Q35. Write a code that inverts a dictionary, swapping keys and values. Ensure that the inverted dictionary correctly handles cases where multiple keys have the same value by storing the keys as a list in the inverted dictionary.
# Ans:-
def invert_dictionary(original_dict):
    inverted_dict = {}
    for key, value in original_dict.items():
        if value in inverted_dict:
            inverted_dict[value].append(key)  # Add key to existing list
        else:
            inverted_dict[value] = [key]  # Create new list with the key
    return inverted_dict

# Example usage:
data = {'a': 1, 'b': 2, 'c': 1, 'd': 3}

inverted = invert_dictionary(data)
print("Inverted Dictionary:", inverted)
```
